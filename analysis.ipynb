# Sales & Customer Data Analysis

This notebook analyzes sales and customer transaction data using Python and Pandas to extract insights that support data-driven business decisions.

```python
import pandas as pd

# Load dataset
df = pd.read_csv("sales_data.csv")

# Preview the data
df.head()
```

## Dataset Overview

In this section, we explore the structure of the dataset and understand the available columns.

```python
df.info()
```

## Basic Sales Statistics

We calculate total sales, average sales value, and identify general patterns in the data.

```python
total_sales = df["Amount"].sum()
average_sales = df["Amount"].mean()

print("Total Sales:", total_sales)
print("Average Sale Amount:", round(average_sales, 2))
```

## Sales by Customer

We aggregate sales data by customer to identify top contributors.

```python
sales_by_customer = df.groupby("Customer")["Amount"].sum()
sales_by_customer
```

## Top Customer Analysis

We identify the customer with the highest total sales.

```python
top_customer = sales_by_customer.idxmax()
top_amount = sales_by_customer.max()

print("Top Customer:", top_customer)
print("Total Sales Amount:", top_amount)
```

## Monthly Sales Analysis

This section analyzes average sales performance across different months.

```python
avg_monthly_sales = df.groupby("Month")["Amount"].mean()
avg_monthly_sales
```

## Business Insights

- A small number of customers contribute a significant portion of total sales.
- Sales performance varies across months, which may indicate seasonal trends.
- Identifying top customers can help prioritize customer retention strategies.
- Aggregated sales analysis supports data-driven business decisions.

## Conclusion

This analysis demonstrates how Python and Pandas can be used to clean, aggregate, and analyze sales data. 
The extracted insights can support strategic decisions related to sales performance and customer management.
